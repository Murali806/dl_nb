## Derivation of Beam Pattern Formula

### Setup - 2 Mic Array

```
Sound Source at angle θ
         \
          \
    Mic1 -------- Mic2
         d
```

Each mic receives the signal $s(t)$ but with a delay:
- **Mic1**: $x_1(t) = s(t)$ → reference mic
- **Mic2**: $x_2(t) = s(t - \tau(\theta))$ → delayed signal

Where delay is:

$$\tau(\theta) = \frac{d \cdot \sin(\theta)}{c}$$
---

### Step 1: Convert to Frequency Domain

Using the Fourier Transform delay property:
$$\mathcal{F}\{s(t - \tau)\} = S(f) \cdot e^{-j2\pi f \tau}$$
So each mic in frequency domain:
$$\ x_1(f) = S(f)$$ $$\ x_2(f) = S(f) \cdot e^{-j2\pi f \tau(\theta)}$$
Substituting $\tau(\theta)$:
$$x_2(f) = S(f) \cdot e^{-j2\pi f \cdot \frac{d \cdot \sin(\theta)}{c}}$$
| Term | Meaning |
|------|---------|
| $f$ | frequency of signal |
| $\tau(\theta)$ | time delay based on angle |
| $f \cdot \tau$ | how many cycles of delay |
| $2\pi f \cdot \tau$ | delay in radians (phase) |
| $e^{-j2\pi f\tau}$ | unit vector rotation in complex plane |

---

### Step 2: Delay-and-Sum Beamformer

To **steer** to angle $\theta_0$, we:
1. **Compensate** Mic2's delay → advance it by $\tau(\theta_0)$
2. **Sum** both mics
3. **Average** (divide by 2)

In time domain:
$$y(t) = \frac{1}{2}\left[x_1(t) + x_2(t + \tau(\theta_0))\right]$$
Advancing by $\tau$ in time domain = multiply by $e^{+j2\pi f \tau}$ in frequency domain:
$$Y(f) = \frac{1}{2}\left[X_1(f) + X_2(f) \cdot e^{+j2\pi f \tau(\theta_0)}\right]$$

Substituting $X_1(f)$ and $X_2(f)$:
$$Y(f) = \frac{1}{2}\left[S(f) + S(f) \cdot e^{-j2\pi f \tau(\theta)} \cdot e^{+j2\pi f \tau(\theta_0)}\right]$$ $$Y(f) = \frac{S(f)}{2}\left[1 + e^{j2\pi f (\tau(\theta_0) - \tau(\theta))}\right]$$

---

### Step 3: Beam Pattern = Gain at each angle

Beam pattern is the gain the beamformer gives to signal from angle $\theta$:
$$B(\theta, f) = \left|\frac{Y(f)}{S(f)}\right|$$
For steering at **$\theta_0 = 0$ (broadside)** → $\tau(\theta_0) = 0$, simplifies to:
$$\boxed{B(\theta, f) = \left|\frac{1 + e^{-j2\pi f \tau(\theta)}}{2}\right|}$$
---

### Step 4: Intuition for different angles

**Case 1: $\theta = \theta_0$ (target direction)**

Both mics are in phase → $\tau = 0$
$$B = \left|\frac{1 + e^{0}}{2}\right| = \left|\frac{1+1}{2}\right| = 1 \quad \checkmark \text{ (maximum gain)}$$
**Case 2: Out of phase direction**

At frequency where $f \cdot \tau = 0.5$ → phase difference = $\pi$
$$B = \left|\frac{1 + e^{-j\pi}}{2}\right| = \left|\frac{1 + (-1)}{2}\right| = 0 \quad \text{(null/rejected)}$$

**Case 3: Any other angle**

Partial cancellation → $B$ between 0 and 1

| Angle | Phase Difference | Result |
|-------|-----------------|--------|
| Target $\theta_0$ | 0 (in phase) | Gain = 1 (max) |
| Opposite | $\pi$ (out of phase) | Gain = 0 (null) |
| Other angles | partial | $0 <$ Gain $< 1$ |

---

### Summary

- Sound from **target direction** → mics in phase → signals add up → **Gain = 1**
- Sound from **other directions** → mics out of phase → signals cancel → **Gain = 0**
- This is why beamforming **amplifies target direction** and **suppresses others!**
