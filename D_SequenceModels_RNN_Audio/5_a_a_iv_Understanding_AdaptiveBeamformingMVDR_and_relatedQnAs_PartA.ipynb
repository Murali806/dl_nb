## 1.1 Linear Algebra Basics

### What is a Vector?

A **vector** is simply a **list of numbers**. Think of it as coordinates or measurements.

**Example**: Microphone signals at two mics
```
x = [0.5, 0.3]  ← Mic 1 recorded 0.5, Mic 2 recorded 0.3
```

### What is a Matrix?

A **matrix** is a **table of numbers** (rows and columns).

**Example**: Signals at 2 mics over 3 time steps
```
     Time 1  Time 2  Time 3
Mic1 [ 0.5    0.6     0.4  ]
Mic2 [ 0.3    0.4     0.2  ]
```

### Visual Representation

````python
def visualize_vectors_and_matrices():
    """
    Visualize vectors and matrices with simple examples.
    """
    fig, axes = plt.subplots(1, 3, figsize=(16, 5))
    
    # Plot 1: Vector as arrow
    ax = axes[0]
    vector = np.array([3, 2])
    ax.arrow(0, 0, vector[0], vector[1], head_width=0.3, head_length=0.3, 
             fc='blue', ec='blue', linewidth=3)
    ax.scatter([0], [0], s=100, c='red', zorder=5, label='Origin')
    ax.scatter([vector[0]], [vector[1]], s=100, c='blue', zorder=5, label='Vector endpoint')
    ax.text(vector[0]/2 - 0.3, vector[1]/2 + 0.3, 'v = [3, 2]', fontsize=14, weight='bold')
    ax.set_xlim(-0.5, 4)
    ax.set_ylim(-0.5, 3)
    ax.set_xlabel('x₁ (First element)', fontsize=11)
    ax.set_ylabel('x₂ (Second element)', fontsize=11)
    ax.set_title('Vector: A List of Numbers (visualized as arrow)', fontsize=12, weight='bold')
    ax.grid(True, alpha=0.3)
    ax.legend()
    ax.set_aspect('equal')
    
    # Plot 2: Matrix as heatmap
    ax = axes[1]
    matrix = np.array([[0.5, 0.6, 0.4],
                       [0.3, 0.4, 0.2]])
    im = ax.imshow(matrix, cmap='Blues', aspect='auto')
    ax.set_xticks([0, 1, 2])
    ax.set_yticks([0, 1])
    ax.set_xticklabels(['Time 1', 'Time 2', 'Time 3'])
    ax.set_yticklabels(['Mic 1', 'Mic 2'])
    ax.set_title('Matrix: A Table of Numbers (2 mics × 3 time steps)', fontsize=12, weight='bold')
    
    # Add values as text
    for i in range(2):
        for j in range(3):
            ax.text(j, i, f'{matrix[i, j]:.1f}', ha='center', va='center', 
                   fontsize=12, weight='bold', color='white')
    
    plt.colorbar(im, ax=ax, label='Signal Value')
    
    # Plot 3: Multiple vectors
    ax = axes[2]
    vectors = np.array([[2, 1], [1, 3], [-1, 2]])
    colors = ['blue', 'red', 'green']
    labels = ['v₁ = [2, 1]', 'v₂ = [1, 3]', 'v₃ = [-1, 2]']
    
    for vec, color, label in zip(vectors, colors, labels):
        ax.arrow(0, 0, vec[0], vec[1], head_width=0.2, head_length=0.2,
                fc=color, ec=color, linewidth=2.5, alpha=0.7, label=label)
    
    ax.scatter([0], [0], s=100, c='black', zorder=5)
    ax.set_xlim(-2, 3)
    ax.set_ylim(-1, 4)
    ax.set_xlabel('x₁', fontsize=11)
    ax.set_ylabel('x₂', fontsize=11)
    ax.set_title('Multiple Vectors (different directions & magnitudes)', fontsize=12, weight='bold')
    ax.grid(True, alpha=0.3)
    ax.legend(loc='upper left', fontsize=9)
    ax.set_aspect('equal')
    
    plt.tight_layout()
    plt.show()
    
    print("\n" + "="*70)
    print("KEY CONCEPTS:")
    print("="*70)
    print("• Vector: A list of numbers [x₁, x₂, ..., xₙ]")
    print("• Matrix: A table of numbers (rows × columns)")
    print("• In beamforming:")
    print("  - Vectors represent signals at different mics")
    print("  - Matrices represent signals over time")
    print("="*70)

visualize_vectors_and_matrices()
````

## Quick Reference

### Key Formulas

1. **Time Delay**: $\tau = \frac{d \cdot \sin(\theta)}{c}$

2. **Phase Shift**: $\phi = 2\pi f \tau$

3. **Complex Representation**: $e^{i\theta} = \cos(\theta) + i\sin(\theta)$

4. **Variance**: $\sigma^2 = \frac{1}{N}\sum_{i=1}^{N}(x_i - \mu)^2$

5. **Covariance**: $\text{cov}(X, Y) = \frac{1}{N}\sum_{i=1}^{N}(x_i - \mu_x)(y_i - \mu_y)$

### Key Concepts

- **Vector**: List of numbers representing signals
- **Matrix**: Table organizing multiple signals
- **Complex Number**: Represents amplitude AND phase
- **Covariance Matrix**: Captures noise statistics across mics
- **Phase Alignment**: Key to beamforming success
