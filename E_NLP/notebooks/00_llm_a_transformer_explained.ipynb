1st step (Token Embedding)

Tokenization splits input text into tokens - small units like words or part of words. GPT-2(small) has 50,257 token vocabulary, each with a unique ID.  In token embedding step, every token is matched to a 768- number vector from a large lookup table. These vectors are learned during training to best represent each token's meaning.
